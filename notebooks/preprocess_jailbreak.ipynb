Preprocessing jailbreak csv

```python
import pandas as pd
import json
```

```python
df = pd.read_csv('../data/raw/SimsonsunJailbreakPrompts.csv')
```

```python
print(df.head())
```

```python
df = df[['Prompt']].dropna()
```

```python
df_sample = df.sample(n=50, random_state=42).copy()
```

```python
df_sample["id"] = [f"id_1_{i+1}" for i in range(len(df_sample))]
df_sample["label"] = "injection" 
df_sample = df_sample[["id", "Prompt", "label"]]
```

```python
output_file = "../data/cleaned/jailbreak_prompts_Simsonsun.jsonl"
with open(output_file, "w", encoding="utf-8") as f:
    for _, row in df_sample.iterrows():
        json_line = {
            "id": row["id"],
            "prompt": row["Prompt"],
            "label": row["label"]
        }
        f.write(json.dumps(json_line, ensure_ascii=False) + "\n")
```

```python
print(f"Random 50 JSONL file created: {output_file}")
```

```python
df2 = pd.read_csv('../data/raw/Bravanskycompact-jailbreaks.csv')
```

```python
df2 = df2.dropna(subset=["attack"])
```

```python
df_sample2 = df2.sample(n=50, random_state=42).copy()
```

```python
df_sample2["id"] = [f"id_2_{i+1}" for i in range(len(df_sample2))]
df_sample2["label"] = "injection"
```

```python
output_file2 = "../data/cleaned/jailbreak_prompts_Bravansky.jsonl"
with open(output_file2, "w", encoding="utf-8") as f:
    for _, row in df_sample2.iterrows():
        json_line = {
            "id": row["id"],
            "prompt": row["attack"],  # lowercase 'prompt'
            "label": row["label"]
        }
        f.write(json.dumps(json_line, ensure_ascii=False) + "\n")
```

```python
rows = []

with open(
    "../data/raw/Prompt_INJECTION_And_Benign_DATASET.jsonl",
    "r",
    encoding="utf-8"
) as f:
    for line in f:
        rows.append(json.loads(line))

df3 = pd.DataFrame(rows)

# Keep only malicious prompts
df3 = df3[df3["label"] == "malicious"]

print("Malicious prompts available:", len(df3))

df_sample3 = df3.sample(
    n=min(50, len(df3)),
    random_state=42
).copy()

df_sample3["id"] = [
    f"id_3_{i+1}"
    for i in range(len(df_sample3))
]

df_sample3["label"] = "injection"

output_file3 = "../data/cleaned/jailbreak_prompts_prompt_injection.jsonl"

with open(output_file3, "w", encoding="utf-8") as f:
    for _, row in df_sample3.iterrows():
        json_line = {
            "id": row["id"],
            "prompt": row["prompt"],
            "label": row["label"]
        }
        f.write(json.dumps(json_line, ensure_ascii=False) + "\n")

print(f"Created: {output_file3}")
```

```python
file1 = "../data/cleaned/jailbreak_prompts_Simsonsun.jsonl"
file2 = "../data/cleaned/jailbreak_prompts_Bravansky.jsonl"
file3 = "../data/cleaned/jailbreak_prompts_prompt_injection.jsonl"

output_merged = "../data/cleaned/150_jailbreak_prompts.jsonl"
```

```python
with open(output_merged, "w", encoding="utf-8") as outfile:
    for fname in [file1, file2, file3]:
        with open(fname, "r", encoding="utf-8") as infile:
            for line in infile:
                outfile.write(line)

print(f"Merged JSONL file created: {output_merged}")
```
